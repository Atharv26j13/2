```python
import imaplib
import email
from email.header import decode_header
from email.utils import parseaddr
import smtplib
from email.message import EmailMessage

# ==========================
# CONFIGURATION
# ==========================
IMAP_SERVER = "imap.gmail.com"
SMTP_SERVER = "smtp.gmail.com"
EMAIL_USER = "temporaryatharv@gmail.com"
EMAIL_PASS = "bfpr jgxi jmhx esnf"

# Define this before running the code
# Example:
# SENDER_EMAIL = "atharvlotusvalley@gmail.com"

recipient_email = SENDER_EMAIL


def read_latest_email():
    global recipient_email

    try:
        mail = imaplib.IMAP4_SSL(IMAP_SERVER)
        mail.login(EMAIL_USER, EMAIL_PASS)

        # Open Inbox
        mail.select("INBOX")

        # Search for emails from SENDER_EMAIL
        status, messages = mail.search(None, f'(FROM "{SENDER_EMAIL}")')

        if status != "OK":
            print("Search failed.")
            return False

        mail_ids = messages[0].split()

        if not mail_ids:
            print(f"No emails found from {SENDER_EMAIL}.")
            return False

        latest_id = mail_ids[-1]

        status, data = mail.fetch(latest_id, "(RFC822)")

        if status != "OK":
            print("Couldn't fetch email.")
            return False

        msg = email.message_from_bytes(data[0][1])

        # Store sender email
        _, recipient_email = parseaddr(msg["From"])

        # Decode subject
        subject = msg["Subject"]
        if subject:
            decoded = decode_header(subject)[0]
            subject = decoded[0]
            if isinstance(subject, bytes):
                subject = subject.decode(decoded[1] or "utf-8", errors="ignore")

        print("=" * 50)
        print("FROM:", recipient_email)
        print("SUBJECT:", subject)
        print("=" * 50)

        body = ""

        if msg.is_multipart():
            for part in msg.walk():
                if (
                    part.get_content_type() == "text/plain"
                    and "attachment" not in str(part.get("Content-Disposition"))
                ):
                    body = part.get_payload(decode=True).decode(errors="ignore")
                    break
        else:
            body = msg.get_payload(decode=True).decode(errors="ignore")

        print(body)

        mail.close()
        mail.logout()

        return True

    except Exception as e:
        print("Error reading email:")
        print(e)
        return False


# ==========================
# READ EMAIL
# ==========================
found = read_latest_email()

print()

# ==========================
# SEND EMAIL
# ==========================
choice = input("Send a message to this person? (y/n): ").strip().lower()

if choice == "y":
  a = int(input("Do you want to\n1)Generate an Email\n2)Write an email"))
  if a==2:
    # If no email was found, still send to SENDER_EMAIL
    if recipient_email is None:
        recipient_email = SENDER_EMAIL

    message = input("\nEnter your message:\n")

    msg = EmailMessage()
    msg["Subject"] = Summarise2(message)
    msg["From"] = EMAIL_USER
    msg["To"] = recipient_email
    msg.set_content(message)

    try:
        with smtplib.SMTP_SSL(SMTP_SERVER, 465) as smtp:
            smtp.login(EMAIL_USER, EMAIL_PASS)
            smtp.send_message(msg)

        print("\n✅ Email sent successfully!")

    except Exception as e:
        print("\n❌ Failed to send email.")
        print(e)
  if a==1:
    Generate()
```
