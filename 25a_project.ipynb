```python
import random
import time

print("Rock!")
time.sleep(1)
print("Paper!")
time.sleep(1)
print("Scissors!")
time.sleep(1)
print("Shoot!")

a=int(input("1,2 or 3?\nRock, Paper or Scissors?"))
b = random.randint(1,3)
if b == 1:
  print("Robot chose Rock!")
if b == 2:
  print("Robot chose Paper!")
if b == 3:
  print("Robot chose Scissors!")
if a == b:
  print("Tie!")
if b == a+1 or b == a-2:
  print("Robot wins!")
if a == b+1 or a == b-2:
  print("You win!")
```
