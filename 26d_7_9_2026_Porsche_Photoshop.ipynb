```python
# @title
import cv2
import numpy as np
import matplotlib.pyplot as plt

print("======== PHOTOSHOP ========")
print("Image uploaded:")

image = cv2.imread("filters_format.jpg")

if image is None:
    print("ERROR: Image could not be found!")
    exit()

# OpenCV loads images as BGR
# We keep the image as BGR internally
def show_image():
    rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    plt.imshow(rgb)
    plt.axis("off")
    plt.show()


show_image()


```

```python
# @title
def options():
    global image

    while True:
        print("\nOptions:")
        print("1. Blur")
        print("2. Edge")
        print("3. Gray")
        print("4. Hue")
        print("5. Exit")

        a = int(input("Enter choice: "))

        if a == 1:
            print("\n1. Gaussian Blur")
            print("2. Normal Blur")

            choice = int(input("Enter choice: "))
            x = int(input("Enter strength: "))

            if x <= 0:
                print("Strength must be greater than 0!")
                continue

            if x % 2 == 0:
                x += 1

            if choice == 1:
                image = cv2.GaussianBlur(image, (x, x), 0)

            elif choice == 2:
                image = cv2.blur(image, (x, x))

            else:
                print("Invalid choice!")
                continue

            show_image()

        elif a == 2:
            print("\n1. Canny")
            print("2. Laplacian")
            print("3. Sobel")

            choice = int(input("Enter choice: "))

            if choice == 1:
                x = int(input("Enter threshold 1: "))
                y = int(input("Enter threshold 2: "))

                gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
                image = cv2.Canny(gray, x, y)

                plt.imshow(image, cmap="gray")
                plt.axis("off")
                plt.show()

            elif choice == 2:
                x = int(input("Enter strength: "))

                if x % 2 == 0:
                    x += 1

                gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
                image = cv2.Laplacian(gray, cv2.CV_64F, ksize=x)
                image = cv2.convertScaleAbs(image)

                plt.imshow(image, cmap="gray")
                plt.axis("off")
                plt.show()

            elif choice == 3:
                x = int(input("Enter strength: "))

                if x % 2 == 0:
                    x += 1

                gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
                image = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=x)
                image = cv2.convertScaleAbs(image)

                plt.imshow(image, cmap="gray")
                plt.axis("off")
                plt.show()

            else:
                print("Invalid choice!")

        elif a == 3:

          if len(image.shape) == 3:
            image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

          plt.imshow(image, cmap="gray")
          plt.axis("off")
          plt.show()

        elif a == 4:
            x = int(input("Enter hue: "))

            if len(image.shape) == 2:
                image = cv2.cvtColor(image, cv2.COLOR_GRAY2BGR)

            hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)

            hsv[:, :, 0] = (hsv[:, :, 0].astype(int) + x) % 180

            image = cv2.cvtColor(hsv, cv2.COLOR_HSV2BGR)

            show_image()

        elif a == 5:
            print("Goodbye!")
            break

        else:
            print("Invalid choice!")


options()
```
