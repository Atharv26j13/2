```python
import cv2
import numpy as np
import math
import vcam
import matplotlib.pyplot as plt

plt.figure(figsize=(20,20))
img = cv2.imread("minions.jpg")
H,W = img.shape[:2]

c1 = vcam.vcam(H=H,W=W)

plane = vcam.meshGen(H,W)

plane.Z += 20*np.exp(-0.5*(((plane.X**1.0/plane.W)/0.1)/(0.1*np.sqrt(2*np.pi))))
pts3d = plane.getPlane()

pts3d = c1.project(pts3d)
map_x,map_y = c1.getMaps(pts3d)

output = cv2.remap(img,map_x,map_y,cv2.INTER_LINEAR)

plt.subplot(1,2,1)
plt.title("Funny Mirror")
plt.imshow(cv2.cvtColor(np.hstack((img,output)),cv2.COLOR_BGR2RGB))
```
